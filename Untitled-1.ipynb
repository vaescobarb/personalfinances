```python
import pandas as pd

df = pd.read_excel('D:/OneDrive/Documentos/PERSONAL FINANCES/dnb_status_files/20250617_siste_transaksjoner.xlsx')
```

```python
# Transform the 'Dato' column to datetime format, keeping only the date part
df['Dato'] = pd.to_datetime(df['Dato'], errors='coerce').dt.date
```

```python
df
```

```python
df_money_manager = pd.read_excel('D:/OneDrive/Documentos/PERSONAL FINANCES/money_manager_files/2025-06-17.xlsx')
```

```python
# Keep only rows from df_money_manager where Konten is DNB
df_money_manager = df_money_manager[df_money_manager['Konten'] == 'DNB']

# In df_money_manager conver Zeitraum to datetime format, keeping only the date part
df_money_manager['Zeitraum'] = pd.to_datetime(df_money_manager['Zeitraum']).dt.date
```

```python
# Make a new money_manager df, keeping only the columns 'Zeitraum', 'Betrag', and 'Kategorie'
df_money_manager = df_money_manager[['Zeitraum', 'Betrag', 'Kategorie']]
```

```python
df_money_manager
```

```python
# Standardize date columns for comparison
df_compare = df.copy()
df_compare['Dato'] = pd.to_datetime(df_compare['Dato'])
df_money_manager_compare = df_money_manager.copy()
df_money_manager_compare['Zeitraum'] = pd.to_datetime(df_money_manager_compare['Zeitraum'])

# Prepare for merge by selecting relevant columns and renaming for consistency
df_subset = df_compare[['Dato', 'Ut fra konto']].dropna().rename(columns={'Dato': 'Date', 'Ut fra konto': 'Amount'})
mm_subset = df_money_manager_compare[['Zeitraum', 'Betrag']].rename(columns={'Zeitraum': 'Date', 'Betrag': 'Amount'})

# Find rows in df not in money_manager
df_not_in_mm = df_subset.merge(mm_subset, on=['Date', 'Amount'], how='left', indicator=True)
df_not_in_mm = df_not_in_mm[df_not_in_mm['_merge'] == 'left_only']

# Find rows in money_manager not in df
mm_not_in_df = mm_subset.merge(df_subset, on=['Date', 'Amount'], how='left', indicator=True)
mm_not_in_df = mm_not_in_df[mm_not_in_df['_merge'] == 'left_only']

# Highlight results
print("Rows in df not in money_manager:")
display(df[df['Dato'].isin(df_not_in_mm['Date']) & df['Ut fra konto'].isin(df_not_in_mm['Amount'])])

print("Rows in money_manager not in df:")
display(df_money_manager[df_money_manager['Zeitraum'].isin(mm_not_in_df['Date']) & df_money_manager['Betrag'].isin(mm_not_in_df['Amount'])])
```
